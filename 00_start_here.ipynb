> being switched on by p53, and losing that switch when p53 is lost.*

## Why this is a real question

`TP53` is the most frequently mutated gene in human cancer. The protein
it encodes, p53, is a **transcription factor**: when a cell is damaged,
p53 switches on a set of target genes that arrest the cell cycle or
trigger cell death.

So when p53 is lost, two different things can happen to a gene:

| if your gene is… | then when p53 is lost… |
|---|---|
| a **direct p53 target** — switched on *by* p53 | it should go **down**, because the switch is gone |
| a **proliferation gene** — held back by the p53 checkpoint | it should go **up**, because the brake is gone |
| **neither** | nothing much should happen |

**You cannot guess which of these your gene is.** You have to measure it.

A result of "no detectable difference" is a **real, valid, fully
markable answer**. Some of you will get one. Some genes that every
textbook calls p53 targets turn out not to respond in actual patient
data — if that is your gene, you have the most interesting result in the
class, not the worst.

## How to build each function

**The function is the end product, not the starting point.** Nobody
writes a working function on the first attempt, and you are not expected
to. Staring at an empty `def` is the slowest way to begin.

Work in three stages:

**1. Experiment with raw code.** Add a cell, load the data, and poke at
it. Print things. Get one line working, then the next. Check what
`df.dtypes` actually says before you assume. This is throwaway code and
nobody marks it — the point is to find out what the data is really like
before you commit to anything.

**2. Break it into small pieces.** Once something works end to end, pull
the steps apart into small functions that each do one job — parsing a
barcode, say, or converting days to years. Short functions are easier to
test, easier to fix, and easier to read six weeks later.

**3. Assemble the required function.** Finally, wrap it all up in the
function the contract asks for, with the exact name and arguments given,
and run the check cell underneath it. Green ticks mean you are done.

> **You may write as many extra functions as you like.** Only the named
> ones are checked, but helpers you chose well are exactly what the
> code-quality marks are looking for — far better than one long function
> that does everything.
